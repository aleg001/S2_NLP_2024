# 1. ¿Qué es wav2vec2?

Es un modelo autosupervisado para aprender representaciones de audio directamente a partir de datos sin etiquetar, lo que reduce la necesidad de grandes cantidades de datos etiquetados para tareas de reconocimiento de voz. Su arquitectura se basa en convoluciones, transformadores y cuantización de representaciones de audio latentes.

# 2. ¿Qué resuelve wav2vec2?

wav2vec2 resuelve la necesidad de grandes cantidades de datos etiquetados para entrenar sistemas de reconocimiento de voz. Entrena un modelo preciso con muy pocos datos etiquetados y cantidades masivas de datos no etiquetados.

# 3. Diferencias con otros enfoques
Wav2vec2 a diferencia de otros, este los combina en un solo proceso de entrenamiento. Resuelve una tarea contrastiva donde el modelo debe identificar la representación correcta dentro de un conjunto de distracciones.

# 4. Arquitectura
wav2vec2 tiene tres componentes principales:

- Codificador de características: Una red de convolución profunda que toma como entrada la señal de audio cruda y produce representaciones latentes.
- Transformador: Se utiliza para generar representaciones contextuales que capturan la información de toda la secuencia de audio.
- Módulo de cuantización: Se utiliza para convertir las representaciones continuas en representaciones discretas mediante la técnica de product quantization y la técnica Gumbel Softmax para obtener una salida diferenciable.
# 5. Entrenamiento autosupervisado
El modelo se entrena en un esquema autosupervisado, es decir, sin necesidad de datos etiquetados. Durante el pre-entrenamiento, el modelo enmascara segmentos del audio y debe predecir las representaciones correctas para esos segmentos enmascarados. Esto es similar al entrenamiento de modelos de lenguaje como BERT en el procesamiento de lenguaje natural. Posteriormente, se puede ajustar con datos etiquetados para tareas específicas como el reconocimiento de voz.

# 6. Cuantización de las características de audio
El modelo incluye una etapa de cuantización de las representaciones latentes que permite convertir las características continuas en discretas. Esta cuantización ayuda a que el modelo aprenda representaciones más robustas y generalizables. El uso de múltiples "codebooks" o grupos de cuantización permite la combinación de diferentes características para obtener representaciones ricas del audio.

```python
import re

#Limpie los textos.
def limpia_texto(texto):
    texto = texto.lower()
    texto = re.sub(r'http\S+|www\S+|https\S+', '', texto, flags=re.MULTILINE)
    texto = re.sub(r'\@\w+', '', texto)
    texto = re.sub(r'[^\w\s]', '', texto)
    texto = re.sub(r'\d+', '', texto)
    texto = re.sub(r'\s+', ' ', texto).strip()
    return texto

```

```python
#Defina una funcion para construir el vocabulario. Esto es todos los caracteres que están en nuestro corpus.
#Ejemplo: {a: 3, b:5, c:7}
#Luego, cambie el espacio " " por "|". Esto es una buena practica, para tener un elemento visible para el espacio.
#Al vocabulario debe agregar "[UNK]" y "[PAD]". Uno es para desconocidos, otro es para en blanco. Estos deben tener asignada
#la longitud del vocabulario.
#Guarde el vocabulario en un json: vocab.json


def construir_vocabulario(dataset):
    """
    Función para construir el vocabulario de un corpus, reemplazar los espacios con '|',
    y agregar los tokens '[UNK]' y '[PAD]'.
    """
    # Inicializar un contador de caracteres
    vocab_counter = Counter()

    # Recorrer los textos del dataset para contar los caracteres
    for example in dataset:
        texto = example['text'].replace(" ", "|")  # Reemplazar los espacios por '|'
        vocab_counter.update(texto)  # Contar los caracteres

    # Ordenar el vocabulario por orden de aparición
    vocabulario = {char: i for i, (char, _) in enumerate(vocab_counter.most_common())}

    # Agregar [UNK] y [PAD] con la longitud actual del vocabulario
    vocabulario["[UNK]"] = len(vocabulario)
    vocabulario["[PAD]"] = len(vocabulario)

    # Guardar el vocabulario en un archivo JSON
    with open('vocab.json', 'w') as vocab_file:
        json.dump(vocabulario, vocab_file, ensure_ascii=False, indent=4)

    print(f"Vocabulario creado y guardado como vocab.json. Total de caracteres: {len(vocabulario)}")
    return vocabulario
```

```python
from transformers import Wav2Vec2Processor

# Cargar el procesador combinando tokenizer y feature extractor
processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-base-960h")

def pipeline_procesamiento(example):

    # Procesar el audio (extraer características)
    audio_input = processor(example["audio"]["array"],
                            sampling_rate=example["audio"]["sampling_rate"],
                            return_tensors="pt",
                            padding=True)

    # Verificar si la máscara de atención está presente, ya que no siempre se genera
    input_values = audio_input.input_values[0]
    attention_mask = audio_input.attention_mask[0] if "attention_mask" in audio_input else None

    # Procesar el texto (tokenización)
    labels = processor.tokenizer(example["text"], return_tensors="pt", padding=True).input_ids

    # Retornar los valores de entrada (input_values), la máscara de atención y las etiquetas (labels)
    return {
        "input_values": input_values,  # Extraer características del audio
        "attention_mask": attention_mask,  # Verificar que exista la máscara de atención
        "labels": labels[0]  # IDs del texto tokenizado
    }

# Probar con un ejemplo del dataset
ejemplo_procesado = pipeline_procesamiento(dataset_filtro_train[0])

# Verificar los resultados
print("Valores de entrada (input_values):", ejemplo_procesado["input_values"])
print("Máscara de atención:", ejemplo_procesado["attention_mask"])  # Puede ser None si no se genera
print("Etiquetas (labels):", ejemplo_procesado["labels"])



```

```python
#Use la siguiente funcion para crear los batches en el formato que el modelo necesita
def prepare_dataset(batch):
    audio = batch["audio"]

    # batched output is "un-batched" to ensure mapping is correct
    batch["input_values"] = processor(audio["array"], sampling_rate=audio["sampling_rate"]).input_values[0]
    batch["input_length"] = len(batch["input_values"])

    with processor.as_target_processor():
        batch["labels"] = processor(batch["text"]).input_ids
    return batch
```

```python
from transformers import Wav2Vec2Processor
import torch
from dataclasses import dataclass
from typing import Any, Dict, List, Union

@dataclass
class DataCollatorCTCWithPadding:
    processor: Wav2Vec2Processor
    padding: Union[bool, str] = True

    def __call__(self, features: List[Dict[str, Union[List[int], torch.Tensor]]]) -> Dict[str, torch.Tensor]:
        # Extraer los input_values
        input_features = [{"input_values": feature["input_values"]} for feature in features]

        # Extraer los labels (input_ids)
        label_features = [{"input_ids": feature["labels"]} for feature in features]

        # Aplicar padding a los input_features
        batch = self.processor.pad(
            input_features,
            padding=self.padding,
            return_tensors="pt",
        )

        # Aplicar padding a los labels
        with self.processor.as_target_processor():
            labels_batch = self.processor.pad(
                label_features,
                padding=self.padding,
                return_tensors="pt",
            )

        # Reemplazar los tokens de padding por -100 para que sean ignorados por el cálculo de pérdida
        labels = labels_batch["input_ids"].masked_fill(labels_batch.attention_mask.ne(1), -100)

        # Añadir los labels al batch
        batch["labels"] = labels

        return batch
```
